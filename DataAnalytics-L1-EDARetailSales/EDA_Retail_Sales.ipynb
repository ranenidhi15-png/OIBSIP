```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("retail_sales.csv")

df.head()
```

```python
print("Dataset Shape:")
print(df.shape)

print("\nColumn Names:")
print(df.columns.tolist())

print("\nData Types:")
print(df.dtypes)

print("\nMissing Values:")
print(df.isnull().sum())
```

```python
# Numerical columns
numeric_columns = df.select_dtypes(include=np.number).columns

print("Numerical Columns:")
print(numeric_columns.tolist())

print("\nDescriptive Statistics:")
display(df[numeric_columns].describe())
```

```python
# Mean
mean_values = df[numeric_columns].mean()

# Median
median_values = df[numeric_columns].median()

# Mode
mode_values = df[numeric_columns].mode().iloc[0]

# Standard Deviation
std_values = df[numeric_columns].std()

# Combine all results
statistics = pd.DataFrame({
    "Mean": mean_values,
    "Median": median_values,
    "Mode": mode_values,
    "Standard Deviation": std_values
})

display(statistics)
```

```python
# Convert Date column to datetime
df["Date"] = pd.to_datetime(df["Date"], errors="coerce")

# Check the result
print(df["Date"].dtype)
print(df["Date"].head())
```

```python
# Create Month column
df["Month"] = df["Date"].dt.to_period("M").astype(str)

# Calculate monthly sales
monthly_sales = df.groupby("Month")["Total Amount"].sum()

# Plot monthly sales trend
plt.figure(figsize=(12, 5))
plt.plot(monthly_sales.index, monthly_sales.values, marker="o")

plt.title("Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Total Sales Amount")
plt.xticks(rotation=45)
plt.grid(True)

plt.show()
```

### Observation

The monthly sales trend shows how total sales changed over time. 
The graph helps identify months with higher and lower sales and reveals 
the overall pattern of sales performance throughout the year.

```python
# Create Quarter column
df["Quarter"] = df["Date"].dt.to_period("Q").astype(str)

# Calculate quarterly sales
quarterly_sales = df.groupby("Quarter")["Total Amount"].sum()

# Plot quarterly sales trend
plt.figure(figsize=(10, 5))
plt.plot(quarterly_sales.index, quarterly_sales.values, marker="o")

plt.title("Quarterly Sales Trend")
plt.xlabel("Quarter")
plt.ylabel("Total Sales Amount")
plt.grid(True)

plt.show()
```

### Observation

The quarterly sales trend shows the change in total sales across different quarters.
It helps identify periods of higher and lower sales performance and provides a
broader view of the business trend over time.

```python
# Count customers by age group
age_group_counts = df["Age Group"].value_counts()

print(age_group_counts)

# Plot age group distribution
plt.figure(figsize=(8, 5))
age_group_counts.plot(kind="bar")

plt.title("Customer Distribution by Age Group")
plt.xlabel("Age Group")
plt.ylabel("Number of Customers")
plt.xticks(rotation=0)

plt.show()
```

### Observation

The age group distribution shows the number of customers belonging to
different age groups. This helps understand which age groups contribute
more to the customer base and can support age-specific marketing strategies.

```python
# Count customers by gender
gender_counts = df["Gender"].value_counts()

print(gender_counts)

# Plot gender distribution
plt.figure(figsize=(7, 5))
gender_counts.plot(kind="bar")

plt.title("Customer Distribution by Gender")
plt.xlabel("Gender")
plt.ylabel("Number of Customers")
plt.xticks(rotation=0)

plt.show()
```

### Observation

The gender distribution shows the number of customers in each gender category.
This provides an overview of the customer base and helps understand the
gender composition of the retail sales dataset.

```python
# Calculate total quantity sold for each product category
top_products = (
    df.groupby("Product Category")["Quantity"]
    .sum()
    .sort_values(ascending=False)
    .head(10)
)

print("Top 10 Best-Selling Products/Categories:")
print(top_products)

# Create bar chart
plt.figure(figsize=(10, 5))
top_products.plot(kind="bar")

plt.title("Top 10 Best-Selling Products/Categories")
plt.xlabel("Product Category")
plt.ylabel("Total Quantity Sold")
plt.xticks(rotation=45)

plt.show()
```

### Observation

The chart shows the top-selling product categories based on total quantity sold.
The categories with higher quantities indicate stronger customer demand and
can help businesses plan inventory and promotional activities.

```python
# Calculate total revenue for each product category
category_revenue = (
    df.groupby("Product Category")["Total Amount"]
    .sum()
    .sort_values(ascending=False)
)

print("Revenue by Product Category:")
print(category_revenue)

# Create bar chart
plt.figure(figsize=(10, 5))
category_revenue.plot(kind="bar")

plt.title("Revenue by Product Category")
plt.xlabel("Product Category")
plt.ylabel("Total Revenue")
plt.xticks(rotation=45)

plt.show()
```

### Observation

The revenue by product category shows which categories generate higher
sales revenue. This information helps identify important revenue-generating
categories and supports inventory and sales planning decisions.

```python
# Select important numerical columns
correlation_data = df[
    ["Age", "Quantity", "Price per Unit", "Total Amount"]
]

# Calculate correlation matrix
correlation_matrix = correlation_data.corr()

# Create heatmap
plt.figure(figsize=(8, 6))

sns.heatmap(
    correlation_matrix,
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Correlation Heatmap of Numerical Variables")
plt.show()
```

### Observation

The correlation heatmap shows the relationship between the numerical
variables in the dataset. Positive values indicate that two variables
tend to increase together, while negative values indicate an inverse
relationship. Values closer to zero indicate a weaker relationship.

```python
# Calculate sales revenue by gender
gender_sales = (
    df.groupby("Gender")["Total Amount"]
    .sum()
    .sort_values(ascending=False)
)

print("Sales Revenue by Gender:")
print(gender_sales)

# Create bar chart
plt.figure(figsize=(8, 5))
gender_sales.plot(kind="bar")

plt.title("Sales Revenue by Gender")
plt.xlabel("Gender")
plt.ylabel("Total Sales Revenue")
plt.xticks(rotation=0)

plt.show()
```

### Observation

The gender-wise sales analysis compares the total sales revenue generated
by different gender groups. This provides an additional view of customer
purchasing behaviour and helps understand differences in revenue contribution.
